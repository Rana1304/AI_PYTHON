```python
tool = [{
    "type": "function",
    "function": {
        "name": "get_movie_actor",
        "parameters": {
            "type": "object",
            "properties": {
                "movie_name": {
                    "type": "string"
                }
            },
            "required": ["movie_name"]
        }
    }
},{
    "type": "function",
    "function": {
        "name": "get_movie_director",
        "parameters": {
            "type": "object",
            "properties": {
                "movie_name": {
                    "type": "string"
                }
            },
            "required": ["movie_name"]
        }
    }
},{
    "type": "function",
    "function": {
        "name": "get_movie_rating",
        "parameters": {
            "type": "object",
            "properties": {
                "movie_name": {
                    "type": "string"
                }
            },
            "required": ["movie_name"]
        }
    }
},{
    "type": "function",
    "function": {
        "name": "get_movie_price",
        "parameters": {
            "type": "object",
            "properties": {
                "movie_name": {
                    "type": "string"
                }
            },
            "required": ["movie_name"]
        }
    }
}]
```

```python
def get_movie_actor(movie_name):
    for movie in movies:
        if movie['name'].lower() == movie_name.lower():
            return movie['actor']
    return 'Not found'

def get_movie_director(movie_name):
    for movie in movies:
        if movie['name'].lower() == movie_name.lower():
            return movie['director']
    return 'Not found'

def get_movie_rating(movie_name):
    for movie in movies:
        if movie['name'].lower() == movie_name.lower():
            return movie['rating']
    return 'Not found'

def get_movie_price(movie_name):
    for movie in movies:
        if movie['name'].lower() == movie_name.lower():
            return movie['price']
    return 'Not found'
```

```python
system_prompt = """You are a Movie Information Assistant.

Your job is to answer questions about movies using the movie data provided to you.

The available movie information contains:

* Movie name
* Actor
* Director
* Rating
* Price/Budget

Rules:

1. Use only the information available in the provided movie database.
2. If the user asks for the actor of a movie, return the actor.
3. If the user asks for the director, return the director.
4. If the user asks for the rating, return the rating.
5. If the user asks for the price or budget, return the price/budget.
6. Movie names should be treated as case-insensitive. For example, "avatar", "Avatar", and "AVATAR" refer to the same movie.
7. If the requested movie does not exist in the database, respond with "Not found."
8. Do not invent movie information that is not present in the database.
9. Keep responses clear, concise, and easy to understand.
10. If the user asks for multiple details, provide all the requested details together.

Example:
User: Who is the actor in Avatar?
Assistant: Sam Worthington

User: Who directed Avatar?
Assistant: James Cameron

User: What is the rating of Avatar?
Assistant: 7.9

User: What is the price of Avatar?
Assistant: $237,000,000

User: Give me all information about Avatar.
Assistant:
Movie: Avatar
Actor: Sam Worthington
Director: James Cameron
Rating: 7.9
Budget: $237,000,000
"""
```

```python
def handle_tool_call(message):
    responses = []
    for tool_call in message.tool_calls:
        if tool_call.function.name == "get_movie_actor":
            arguments = json.loads(tool_call.function.arguments)
            actor = arguments.get("movie_name")
            actor_name = get_movie_actor(actor)
            responses.append({
                "role": "tool",
                "content": actor_name,
                "tool_call_id": tool_call.id
            })
        if tool_call.function.name == "get_movie_director":
            arguments = json.loads(tool_call.function.arguments)
            director = arguments.get("movie_name")
            director_name = get_movie_director(director)
            responses.append({
                "role": "tool",
                "content": director_name,
                "tool_call_id": tool_call.id
            })
        if tool_call.function.name == "get_movie_rating":
            arguments = json.loads(tool_call.function.arguments)
            rating = arguments.get("movie_name")
            movie_rating = get_movie_rating(rating)
            responses.append({
                "role": "tool",
                "content": str(movie_rating),
                "tool_call_id": tool_call.id
            })
        if tool_call.function.name == "get_movie_price":
            arguments = json.loads(tool_call.function.arguments)
            price = arguments.get("movie_name")
            movie_price = get_movie_price(price)
            responses.append({
                "role": "tool",
                "content": str(movie_price),
                "tool_call_id": tool_call.id
            })
    return responses
```

```python
def chat(message,history):
    history = [{'role':h['role'],'content':h['content']} for h in history]
    messages = [{'role':'system','content':system_prompt}]+history+[{'role':'user','content':message}]
    response = myAI.chat.completions.create(
        model = model,
        messages = messages,
        tools = tool
    )

    while response.choices[0].finish_reason == 'tool_calls':
        message = response.choices[0].message
        responses = handle_tool_call(message)
        messages.append(message)
        messages.extend(responses)
        response = myAI.chat.completions.create(
                model = model,
                messages = messages,
                tools = tool
            )

    return response.choices[0].message.content

ui.ChatInterface(fn=chat).launch(share=True,inline=False,inbrowser = True)
```
